```python
def decimal_para_binario(numero):
    if numero < 0:
        return "-" + bin(abs(numero))[2:]
    return bin(numero)[2:]


def binario_para_decimal(binario):
    if binario.startswith("-"):
        return -int(binario[1:], 2)
    return int(binario, 2)


def binario_valido(valor):
    valor = valor.strip()

    if valor.startswith("-"):
        valor = valor[1:]

    return valor != "" and all(c in "01" for c in valor)


def decimal_valido(valor):
    valor = valor.strip()

    if valor.startswith("-"):
        valor = valor[1:]

    return valor.isdigit()


def menu():
    while True:
        print("OPÇÕES")
        print("1: Decimal → Binário")
        print("2: Binário → Decimal")
        print("3: Somar binários")
        print("4: Subtrair binários")
        print("5: Multiplicar binários")
        print("0: Sair")

        opcao = input("Escolha uma opção: ").strip()

        # Decimal para binario
        if opcao == "1":
            valor = input("Digite um número decimal: ")

            if not decimal_valido(valor):
                print("Erro: digite apenas números inteiros.")
                continue

            numero = int(valor)

            print("Resultado:", decimal_para_binario(numero))
            break

        # Binario para decimal
        elif opcao == "2":
            valor = input("Digite um número binário: ")

            if not binario_valido(valor):
                print("Erro: digite apenas 0 e 1.")
                continue

            print("Resultado:", binario_para_decimal(valor))
            break

        # Operações
        elif opcao in ["3", "4", "5"]:

            n1 = input("Digite o primeiro número binário: ")
            n2 = input("Digite o segundo número binário: ")

            if not binario_valido(n1) or not binario_valido(n2):
                print("Erro: digite apenas números binários válidos.")
                continue

            decimal1 = binario_para_decimal(n1)
            decimal2 = binario_para_decimal(n2)

            # soma
            if opcao == "3":
                resultado = decimal1 + decimal2
                print("Resultado:", decimal_para_binario(resultado))
                break

            # subtracao
            elif opcao == "4":
                resultado = decimal1 - decimal2
                print("Resultado:", decimal_para_binario(resultado))
                break

            # Multiplicaçao
            elif opcao == "5":
                resultado = decimal1 * decimal2
                print("Resultado:", decimal_para_binario(resultado))
                break

        elif opcao == "0":
            print("Saindo")
            break

        else:
            print("Inválido.")


menu()

# algumas partes eu fiz com IA, apenas para fins de estudo proprio
```
